```python
#Task1
import numpy as np

ratings = np.array([
    [4, 5, 3, 4, 5],
    [5, 4, 4, 3, 5],
    [3, 5, 4, 5, 4],
    [4, 3, 5, 4, 3]
])

print("Ratings:")
print(ratings)

result = ratings[1:3]

print("Ratings of second and third users:")
print(result)
```

```python
#Task2
import numpy as np

steps = np.array([7500, 8200, 9000, 6500, 10000, 7800, 8500, 9200, 7000, 8100])

result = steps[steps > 8000]

print("Steps greater than 8000:")
print(result)
```

```python
#Task3
import numpy as np

scores = np.array([180, 210, 165, 195, 220, 175, 205, 190])

result = scores[[1, 4, 6]]

print("Scores of matches 2, 5 and 7:")
print(result)
```

```python
#Task4
import numpy as np

prices = np.array([1000, 2500, 5000, 1200, 3000])

discounted_prices = prices * 0.90

print("Original prices:")
print(prices)

print("Prices after 10% discount:")
print(discounted_prices)
```
