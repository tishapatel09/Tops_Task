```python
#Task1
import numpy as np

friend1 = np.array([5000, 7000, 8000, 6000, 9000, 10000, 7500])
friend2 = np.array([4500, 6500, 8500, 7000, 8000, 9500, 7000])

addition = friend1 + friend2
subtraction = friend1 - friend2
multiplication = friend1 * friend2
division = friend1 / friend2

print("Friend 1 Steps:", friend1)
print("Friend 2 Steps:", friend2)

print("Addition:", addition)
print("Subtraction:", subtraction)
print("Multiplication:", multiplication)
print("Division:", division)
```

```python
#Task2
import numpy as np

user_preferences = np.array([
    [5, 4, 3],
    [2, 5, 4],
    [4, 3, 5]
])

song_popularity = np.array([
    [2, 3, 4],
    [1, 5, 2],
    [3, 2, 5]
])

recommendation_dot = np.dot(user_preferences, song_popularity)

recommendation_matmul = np.matmul(user_preferences, song_popularity)

print("Recommendation Matrix using dot():")
print(recommendation_dot)

print("\nRecommendation Matrix using matmul():")
print(recommendation_matmul)
```

```python
#Task3
import numpy as np

image = np.array([
    [5, 10, 15, 20],
    [25, 30, 35, 40],
    [45, 50, 55, 60],
    [65, 70, 75, 80]
])

rotated_image = image.T

mean = np.mean(image)
median = np.median(image)
standard_deviation = np.std(image)
variance = np.var(image)

print("Original Image:")
print(image)

print("\nTransposed Image:")
print(rotated_image)

print("\nMean:", mean)
print("Median:", median)
print("Standard Deviation:", standard_deviation)
print("Variance:", variance)
```

```python
#Task4
import numpy as np

rating_matrix = np.array([
    [4, 1, 2],
    [1, 5, 1],
    [2, 1, 3]
])

inverse = np.linalg.inv(rating_matrix)

determinant = np.linalg.det(rating_matrix)

eigenvalues, eigenvectors = np.linalg.eig(rating_matrix)

print("Rating Matrix:")
print(rating_matrix)

print("\nInverse:")
print(inverse)

print("\nDeterminant:")
print(determinant)

print("\nEigenvalues:")
print(eigenvalues)

print("\nEigenvectors:")
print(eigenvectors)
```

```python
#Task5
import numpy as np

orders = np.array([
    [100, 120, 140, 160, 180, 200],
    [90, 110, 130, 150, 170, 190]
])

print("Original Array:")
print(orders)

reshaped = orders.reshape(3, 4)
print("\nReshaped Array (3x4):")
print(reshaped)

flattened = orders.flatten()
print("\nFlattened Array:")
print(flattened)

part1, part2 = np.split(flattened, 2)

print("\nPart 1:")
print(part1)

print("\nPart 2:")
print(part2)

stacked = np.vstack((part1, part2))

print("\nVertically Stacked Array:")
print(stacked)
```
