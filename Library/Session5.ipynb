```python
# Task4

import pandas as pd
from pathlib import Path



base_path = Path.cwd()


orders_path = base_path / "S5_orders.csv"
users_path = base_path / "S5_users.csv"

orders = pd.read_csv(orders_path)
users = pd.read_csv(users_path)


combined = pd.merge(
    orders,
    users,
    on="user_id"
)


print(combined)
```

```python
# Task5

import pandas as pd


today_orders = pd.DataFrame({
    "order_id": [101, 102, 103],
    "item": ["Pizza", "Burger", "Pasta"],
    "price": [250, 180, 220]
})


yesterday_orders = pd.DataFrame({
    "order_id": [104, 105, 106],
    "item": ["Sandwich", "Fries", "Coffee"],
    "price": [150, 100, 120]
})


combined_orders = pd.concat(
    [today_orders, yesterday_orders]
)


combined_orders = combined_orders.reset_index(drop=True)

print(combined_orders)
```
