```python
#Task2
import pandas as pd

df = pd.read_csv("S5_orders.csv")

df["order_date"] = pd.to_datetime(df["order_date"])

df["year"] = df["order_date"].dt.year
df["month"] = df["order_date"].dt.month
df["weekday"] = df["order_date"].dt.day_name()

print(df)
```

```python
#Task3
import pandas as pd

df = pd.read_csv("S5_orders.csv")

df["order_date"] = pd.to_datetime(df["order_date"])

df = df.set_index("order_date")

weekly_orders = df.resample("W").size()

print("Total Orders Per Week:")
print(weekly_orders)
```

```python
#Task4
import pandas as pd

data = {
    "posted_at": [
        "2024-06-01 10:00:00",
        "2024-06-01 12:30:00",
        "2024-06-02 08:15:00",
        "2024-06-02 15:45:00",
        "2024-06-03 09:00:00"
    ]
}

df = pd.DataFrame(data)

df["posted_at"] = pd.to_datetime(df["posted_at"], utc=True)

df["posted_at_kolkata"] = df["posted_at"].dt.tz_convert("Asia/Kolkata")

print(df["posted_at_kolkata"].head())
```

```python
#Task5
import pandas as pd

data = {
    "order_date": [
        "2024-06-01 14:30:00",
        "2024-06-02 09:15:00",
        "2024-06-03 20:45:00",
        "2024-06-08 11:00:00",
        "2024-06-10 16:30:00"
    ]
}

df = pd.DataFrame(data)

df["order_date"] = pd.to_datetime(df["order_date"])

df["is_weekend"] = df["order_date"].dt.dayofweek >= 5

print(df)
```
