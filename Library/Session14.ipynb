```python
#Task2
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "rating": [
        5, 4, 5, 3, 2,
        5, 4, 1, 3, 5,
        4, 5, 2, 3, 4,
        5, 1, 4, 5, 3
    ],
    "category": [
        "Mobile", "Laptop", "Mobile", "Shoes", "Mobile",
        "Laptop", "Shoes", "Mobile", "Laptop", "Mobile",
        "Shoes", "Laptop", "Mobile", "Shoes", "Laptop",
        "Mobile", "Shoes", "Laptop", "Mobile", "Shoes"
    ]
}

df = pd.DataFrame(data)

sns.countplot(data=df, x="rating")

plt.title("Number of Flipkart Reviews by Rating")
plt.xlabel("Rating (Stars)")
plt.ylabel("Number of Reviews")

plt.show()
```

```python
#Task3
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "restaurant": [
        "Restaurant A", "Restaurant B", "Restaurant C",
        "Restaurant D", "Restaurant E", "Restaurant F",
        "Restaurant G", "Restaurant H", "Restaurant I",
        "Restaurant J"
    ],
    "average_cost_for_two": [
        300, 500, 800, 250, 1000,
        400, 700, 350, 1200, 600
    ],
    "user_rating": [
        3.5, 4.0, 4.5, 3.2, 4.7,
        3.8, 4.3, 3.6, 4.8, 4.1
    ]
}

df = pd.DataFrame(data)

sns.scatterplot(
    data=df,
    x="average_cost_for_two",
    y="user_rating"
)

plt.title("Restaurant Cost vs User Rating")
plt.xlabel("Average Cost for Two (₹)")
plt.ylabel("User Rating")

plt.show()
```

```python
#Task4
import pandas as pd

data = {
    "movie": [
        "Movie A", "Movie B", "Movie C",
        "Movie D", "Movie E", "Movie F",
        "Movie G", "Movie H"
    ],
    "genre": [
        "Action", "Comedy", "Action", "Drama",
        "Comedy", "Drama", "Action", "Drama"
    ],
    "box_office_collection": [
        150, 80, 200, 100,
        120, 140, 180, 110
    ]
}

df = pd.DataFrame(data)

average_collection = (
    df.groupby("genre")["box_office_collection"]
      .mean()
      .sort_values(ascending=False)
)

print("Average Box Office Collection by Genre:")
print(average_collection)
```
