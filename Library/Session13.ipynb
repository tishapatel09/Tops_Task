```python
#Task1
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")

sns.pairplot(tips)

plt.show()
```

```python
#Task2
import seaborn as sns
import matplotlib.pyplot as plt

flights = sns.load_dataset("flights")

flight_data = flights.pivot(
    index="month",
    columns="year",
    values="passengers"
)

sns.heatmap(
    flight_data,
    annot=True,
    fmt="d",
    cmap="YlGnBu"
)

plt.title("Number of Passengers by Month and Year")
plt.xlabel("Year")
plt.ylabel("Month")

plt.show()
```

```python
#Task3
import seaborn as sns
import matplotlib.pyplot as plt

fmri = sns.load_dataset("fmri")

sns.relplot(
    data=fmri,
    x="timepoint",
    y="signal",
    hue="event",
    kind="line"
)

plt.title("FMRI Signal Changes Over Time")
plt.xlabel("Time Point")
plt.ylabel("Signal")

plt.show()
```

```python
#Task4
import seaborn as sns
import matplotlib.pyplot as plt

titanic = sns.load_dataset("titanic")

sns.catplot(
    data=titanic,
    x="class",
    y="survived",
    kind="bar",
    errorbar=("ci", 95)
)

plt.title("Titanic Survival Rate by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Survival Rate")

plt.show()
```

```python
#Task5
import seaborn as sns
import matplotlib.pyplot as plt

penguins = sns.load_dataset("penguins")

sns.jointplot(
    data=penguins,
    x="bill_length_mm",
    y="flipper_length_mm",
    kind="reg"
)

plt.show()
```
