```python
#Task1
import pandas as pd

data = {
    "restaurant_name": ["Dominos", "McDonalds", "Dominos", "KFC", "Dominos", "KFC"],
    "order_date": ["2026-09-01", "2026-09-02", "2026-09-01",
                   "2026-09-03", "2026-09-01", "2026-09-03"],
    "amount": [250, 300, 250, 450, 250, 450]
}

df = pd.DataFrame(data)

duplicates = df[df.duplicated(
    subset=["restaurant_name", "order_date"],
    keep=False
)]

print("Duplicate Orders:")
print(duplicates)
```

```python
#Task2
import pandas as pd

reviews = [
    "Good product",
    "Excellent quality",
    "Good product",
    "Fast delivery",
    "Good product",
    "Excellent quality",
    "Average product",
    "Fast delivery",
    "Good product"
]

df = pd.DataFrame({"review": reviews})

review_counts = df["review"].value_counts()

print("Top 3 Most Common Reviews:")
print(review_counts.head(3))
```

```python
#Task3
import pandas as pd

data = {
    "playlist_name": [
        "Energetic Beats",
        "Chill Vibes",
        "Energetic Beats",
        "Midnight Jazz",
        "Chill Vibes"
    ],
    "creator_username": [
        "Tisha",
        "Vaidi",
        "Tisha",
        "Jiya",
        "Vaidi"
    ]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

cleaned_df = df.drop_duplicates()

print("\nCleaned DataFrame:")
print(cleaned_df)
```

```python
#Task4
import pandas as pd

data = {
    "username": [
        "insta_queen",
        "insta-queen",
        "instaqueen",
        "tishhaaa09",
        "insta_queen"
    ]
}

df = pd.DataFrame(data)

df["username"] = df["username"].replace({
    "insta_queen": "instaqueen",
    "insta-queen": "instaqueen",
    "instaqueen": "instaqueen"
})

print(df)
```

```python
#Task5
import pandas as pd

data = {
    "payment_status": [
        "Yes",
        "yes ",
        " Y",
        "No",
        " no ",
        "N",
        " YES "
    ]
}

df = pd.DataFrame(data)

df["payment_status"] = df["payment_status"].str.strip().str.lower()

df["payment_status"] = df["payment_status"].replace({
    "yes": 1,
    "y": 1,
    "no": 0,
    "n": 0
})

print(df)
```
