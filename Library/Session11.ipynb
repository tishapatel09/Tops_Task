```python
#Task1
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr"]
sales = [80, 150, 100, 180]

products = ["A", "B", "C", "D"]
quantity = [28, 35, 25, 40]

x = [1, 2, 3, 4, 5]
y = [10, 25, 15, 30, 20]

labels = ["Food", "Travel", "Shopping", "Other"]
values = [40, 25, 20, 15]

fig, axes = plt.subplots(2, 2, figsize=(10, 8))

axes[0, 0].plot(months, sales, marker="o")
axes[0, 0].set_title("Monthly Sales")
axes[0, 0].set_xlabel("Month")
axes[0, 0].set_ylabel("Sales")

axes[0, 1].bar(products, quantity)
axes[0, 1].set_title("Product Quantity")
axes[0, 1].set_xlabel("Product")
axes[0, 1].set_ylabel("Quantity")

axes[1, 0].scatter(x, y, marker="o")
axes[1, 0].set_title("Scatter Plot")
axes[1, 0].set_xlabel("X Values")
axes[1, 0].set_ylabel("Y Values")

axes[1, 1].pie(values, labels=labels, autopct="%1.1f%%")
axes[1, 1].set_title("Category Distribution")

plt.tight_layout()
plt.show()
```

```python
#Task2
import matplotlib.pyplot as plt

services = ["Zomato", "Swiggy", "Domino's"]
delivery_time = [25, 20, 40]

bars = plt.bar(
    services,
    delivery_time,
    color=["red", "blue", "green"],
    edgecolor="black",
    linewidth=2
)

bars[0].set_hatch("/")
bars[1].set_hatch("\\")
bars[2].set_hatch("x")

plt.title("Average Delivery Time Comparison")
plt.xlabel("Food Delivery Service")
plt.ylabel("Average Delivery Time (minutes)")

plt.show()
```

```python
#Task3
import matplotlib.pyplot as plt

influencers = ["A", "B", "C", "D", "E"]

followers = [100000, 150000, 200000, 175000, 250000]
daily_posts = [3, 5, 4, 6, 7]

fig, ax1 = plt.subplots(figsize=(10, 6))

ax1.plot(
    influencers,
    followers,
    color="blue",
    marker="o",
    linewidth=2,
    label="Followers"
)

ax1.set_xlabel("Influencers")
ax1.set_ylabel("Followers", color="blue")
ax1.tick_params(axis="y", labelcolor="blue")

ax2 = ax1.twinx()

ax2.plot(
    influencers,
    daily_posts,
    color="red",
    marker="s",
    linewidth=2,
    label="Daily Posts"
)

ax2.set_ylabel("Average Daily Posts", color="red")
ax2.tick_params(axis="y", labelcolor="red")

plt.title("Instagram Followers vs Average Daily Posts")

plt.show()
```

```python
#Task4
import matplotlib.pyplot as plt

movies = [
    "Movie A",
    "Movie B",
    "Movie C",
    "Movie D",
    "Movie E"
]

tickets_sold = [18, 16, 27, 13, 30] 
imdb_rating = [7.2, 8.0, 7.5, 6.8, 8.3]

plt.scatter(tickets_sold, imdb_rating, s=100)

for i in range(len(movies)):
    plt.annotate(
        movies[i],
        (tickets_sold[i], imdb_rating[i]),
        xytext=(5, 5),
        textcoords="offset points"
    )

plt.title("Bollywood Movies: Tickets Sold vs IMDb Rating")
plt.xlabel("Tickets Sold (Lakhs)")
plt.ylabel("IMDb Rating")

plt.grid(True)

plt.show()
```

```python
#Task5
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [120, 150, 135, 180, 160, 300]

fig, ax = plt.subplots(figsize=(9, 5))

ax.plot(
    months,
    sales,
    marker="o",
    linewidth=2
)

max_sales = max(sales)
max_index = sales.index(max_sales)

ax.annotate(
    "Big Billion Days",
    xy=(months[max_index], max_sales),
    xytext=(months[max_index], max_sales + 40),
    arrowprops=dict(arrowstyle="->"),
    fontsize=12
)

ax.set_title("Flipkart Monthly Sales")
ax.set_xlabel("Month")
ax.set_ylabel("Sales (₹ Lakhs)")

ax.grid(True)

plt.show()
```
