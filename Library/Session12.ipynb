```python
#Task1
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(10)
delivery_times = np.random.randint(20, 61, 50)

sns.histplot(delivery_times, bins=8, kde=True)

plt.title("Zomato Delivery Time Distribution")
plt.xlabel("Delivery Time (Minutes)")
plt.ylabel("Number of Orders")

plt.show()
```

```python
#Task2
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")

sns.boxplot(
    data=tips,
    x="day",
    y="total_bill"
)

plt.title("Total Bill Distribution by Day")
plt.xlabel("Day of the Week")
plt.ylabel("Total Bill Amount")

plt.show()
```

```python
#Task3
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

sns.set_theme(style="darkgrid")

teams = [
    "CSK", "MI", "RCB", "KKR",
    "GT", "RR", "SRH", "DC"
]

np.random.seed(20)

data = []

for team in teams:
    scores = np.random.randint(120, 221, 10)

    for score in scores:
        data.append([team, score])

df = pd.DataFrame(data, columns=["Team", "Runs"])

sns.violinplot(
    data=df,
    x="Team",
    y="Runs"
)

plt.title("IPL Run Distribution by Team")
plt.xlabel("Team")
plt.ylabel("Runs")

plt.show()
```

```python
#Task4
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

genres = [
    "Pop", "Rock", "Hip-Hop", "Pop", "EDM",
    "Rock", "Pop", "Hip-Hop", "EDM", "Pop",
    "Rock", "Jazz", "Pop", "Hip-Hop", "EDM",
    "Rock", "Pop", "Jazz", "Hip-Hop", "EDM",
    "Pop", "Rock", "Pop", "Hip-Hop", "Jazz",
    "EDM", "Rock", "Pop", "Hip-Hop", "EDM",
    "Pop", "Rock", "Jazz", "Hip-Hop", "Pop",
    "EDM", "Rock", "Pop", "Hip-Hop", "Jazz"
]

df = pd.DataFrame({"Genre": genres})

sns.countplot(
    data=df,
    x="Genre"
)

plt.title("Number of Spotify Songs by Genre")
plt.xlabel("Music Genre")
plt.ylabel("Number of Songs")

plt.show()
```

```python
#Task5
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid")

steps = [
    6500,
    8200,
    7500,
    9000,
    6800,
    10500,
    9500
]

sns.kdeplot(
    steps,
    color="blue",
    fill=True
)

plt.title("Distribution of Daily Step Counts")
plt.xlabel("Number of Steps")
plt.ylabel("Density")

plt.show()
```
