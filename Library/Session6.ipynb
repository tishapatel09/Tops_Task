```python
# Task1

import pandas as pd


ipl = pd.read_csv("S6_ipl_players.csv")

print(ipl)

print("\nMissing values in each column:")
print(ipl.isnull().sum())


print("\nNon-missing values in each column:")
print(ipl.notnull().sum())
```

```python
# Task2

print("Shape before dropping missing values:", ipl.shape)

ipl_clean = ipl.dropna(axis=0, how="any")

print("Shape after dropping missing values:", ipl_clean.shape)

print("\nCleaned DataFrame:")
print(ipl_clean)
```

```python
# Task3

runs_mean = ipl["runs"].mean()

print("Mean of runs:", runs_mean)

ipl["runs"] = ipl["runs"].fillna(runs_mean)

print("\nUpdated runs column:")
print(ipl["runs"])
```

```python
# Task4

import pandas as pd

restaurants = pd.DataFrame({
    "name": ["Dominos", "Pizza Hut", "Subway", "McDonalds", "Burger King", "La Pinoz"],
    "rating": [None, 4.5, None, 4.2, None, 4.4]
})

print("Before filling:")
print(restaurants)

# # Forward fill
restaurants["rating"] = restaurants["rating"].ffill()

# Backward fill remaining missing values
restaurants["rating"] = restaurants["rating"].bfill()

print("\nAfter filling:")
print(restaurants)
```

```python
# Task5

import pandas as pd

ipl = pd.DataFrame({
    "player_name": ["Virat", "Rohit", "Gill", "Dhoni", "Rahul"],
    "runs": [726, None, 500, None, 450],
    "matches": [15, 14, 17, 14, 13]
})

print("IPL Data:")
print(ipl)

print("\nMissing values:")
print(ipl.isnull().sum())

# The missing values in the runs column can be considered MCAR (Missing Completely At Random) because the missing runs are assumed to occur randomly and are not related to the player's name, team, or number of matches. There is no specific pattern explaining why these runs are missing.

```
