```python
# Task2

def name(songs):
    for i in songs:
        yield i

songs = ["tum ho","darkhast","khat","chasni","aarzu"]

for s in name(songs):
    print(s)
```

```python
# Task5

def order_id_generator():
    order_id = 1001
    while True:
        yield order_id
        order_id += 1

orders = order_id_generator()
```
