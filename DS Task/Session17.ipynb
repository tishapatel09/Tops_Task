```python
# 3.Use the datetime module to get the current date and time, then format and print it as 'DD-MM-YYYY HH:MM:SS', similar to how WhatsApp shows message timestamps.Use strftime() to format the output.

from datetime import datetime
current_time = datetime.now()

formatted = current_time.strftime("%d-%m-%Y %H:%M:%S")
print(formatted)
```
