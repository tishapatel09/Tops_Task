Write a Python program for a Food Delivery platform with the following requirements:

1. Fetch the first 5 records from https://jsonplaceholder.typicode.com/posts?_limit=5 and treat them as mock restaurants. Print each record's id and title.
2. Apply a tiered commission to a hardcoded list of 6 order amounts:

   * 10% for orders up to Rs 200
   * 15% for orders from Rs 201 to Rs 500
   * 20% for orders above Rs 500
     Use map() and a lambda, and print each original order amount alongside its calculated commission.
3. Handle API failure by checking the HTTP status code. If the status code is not 200, print a descriptive error message and stop execution without an unhandled exception.
4. Save the 5 fetched records to data/processed/records.json using json.dump() inside a with open() context manager, with indent=2.
5. Keep the code beginner-friendly and easy to understand.


```python
# AI's Original Code

import requests
import json

url = "https://jsonplaceholder.typicode.com/posts?_limit=5"

response = requests.get(url)

if response.status_code != 200:
    print("API request failed. Status code:", response.status_code)
else:
    records = response.json()


print("Mock Restaurants:")
for record in records:
    print("ID:", record["id"], "Title:", record["title"])

orders = [100, 200, 300, 500, 600, 1000]

commission = map(
    lambda amount: amount * 0.10 if amount <= 200
    else amount * 0.15 if amount <= 500
    else amount * 0.20,
    orders
)

print("\nOrder Commission:")
for amount, result in zip(orders, commission):
    print("Order:", amount, "Commission:", result)

with open("data/processed/records.json", "w") as file:
    json.dump(records, file)

```

```python
# My corrected Version

import requests
import json

url = "https://jsonplaceholder.typicode.com/posts?_limit=5"

response = requests.get(url)

if response.status_code != 200:
    print("API request failed. Status code:", response.status_code)
    exit()
else:
    records = response.json()


print("Mock Restaurants:")
for record in records:
    print("ID:", record["id"], "Title:", record["title"])

orders = [100, 200, 300, 500, 600, 1000]

commission = map(
    lambda amount: amount * 0.10 if amount <= 200
    else amount * 0.15 if amount <= 500
    else amount * 0.20,
    orders
)

print("\nOrder Commission:")
for amount, result in zip(orders, commission):
    print("Order:", amount, "Commission:", result)

with open("data/processed/records.json", "w") as file:
    json.dump(records, file, indent=2)

```

However, I found that json.dump() did not include the indent parameter, which made the records.json file difficult to read. I fixed this by adding indent=2 to json.dump(). This makes the JSON file properly formatted and readable.
