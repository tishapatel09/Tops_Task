```python
# Task1

apps = ["Zomato", "Swiggy", "Uber Eats", "Blinkit", "Zepto"]

for app in apps:
    print(app)
```

```python
# Task2

steps = [6500, 8200, 9800, 10500, 12000, 9000, 11000]

day = 0

while day < len(steps):
    if steps[day] > 10000:
        print("First day crossing 10,000 steps:", day + 1)
        break
    day += 1
```

```python
# Task3

def print_teams(teams):
    for team in teams:
        if len(team) <= 6:
            continue
        print(team)

ipl_teams = ["MI", "CSK", "RCB", "Hydrabad", "Punjab", "Gujarat", "Lucknow"]

print_teams(ipl_teams)
```

```python
# Task4

durations = [210, 180, 240, 195]

for position, duration in enumerate(durations, start=1):
    print(f"Song {position}: {duration} seconds")
```

```python
# Task5

prices = [500, 700, 0, 600, 400, 900]

total = 0

for price in prices:
    if price == 0:
        continue

    total += price

    if total > 2000:
        break

print("Final Total: ₹", total)
```
