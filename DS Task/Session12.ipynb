```python
# Task4

def format_followers(num):
    if num >= 1000000:
        return f"{num/1000000:g}M"
    elif num >= 1000:
        return f"{num/1000:g}K"
    else:
        return str(num)

followers = [950, 1500, 25000, 1200000]

list(map(format_followers, followers))

```
