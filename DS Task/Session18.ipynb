```python
# 2.Write a Python function using re.search() that checks if a string contains a valid date in the format 'DD/MM/YYYY'. The function should return True if a date is found, otherwise False.

import re

def check_date(text):
    pattern = r"\b\d{2}/\d{2}/\d{4}\b"
    
    if re.search(pattern, text):
        return True
    else:
        return False


# print(check_date("My birthday is 25/12/2000"))
print(check_date("Today is 5/8/2026"))


```

```python
# 3. Given a block of text containing multiple prices (like 'Rs. 299', 'Rs. 1500', etc.), use re.findall() to extract all the numeric price values as integers and print their sum.Look for patterns like 'Rs. ' followed by one or more digits.

text = "here is a list of prices like Rs. 299 and Rs. 1500 and Rs. 670" 

prices = re.findall(r"\b\w{2}. (\d+)", text)

prices = [int(price) for price in prices]

print(prices)
sum(prices)

```

```python
# 4.Use re.sub() to replace all email addresses in a string with '[hidden email]' and print the modified string. Do not use any external libraries except re.

text = "Contact us at abc@gmail.com or support@yahoo.com for help."

pattern = r'\b[\w]+@[\w]+\.\w+\b'

result = re.sub(pattern, '[hidden email]', text)

print(result)
```

```python
# 5. Download a sample Instagram comments text file (or create your own with at least 10 lines), then write a Python script to extract all valid Instagram usernames (pattern: starts with '@', followed by letters, numbers, underscores, minimum 3 characters) using re.findall() and print the unique usernames.

with open("cmt.txt", "r", encoding="utf-8") as file:
    text = file.read()

pattern = r'@[A-Za-z0-9_]{2,}'

usernames = re.findall(pattern, text)

unique_usernames = set(usernames)

print("Unique usernames:")

for username in unique_usernames:
    print(username)
```
